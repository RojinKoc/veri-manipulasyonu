```python
df = pd.DataFrame({'gruplar':['A','B','C','A','B','C'],
                  'degisken1':[10,23,33,22,11,99],
                  'degisken2':[100,253,333,262,111,969]},
                 columns = ['gruplar','degisken1','degisken2'])
df
```

```python
df.groupby("gruplar").mean()
```

```python
df.groupby("gruplar").aggregate(["min", "median", "max"])
# Aggregate: Birden fazla istatistiksel fonksiyonu aynı anda uygulamak için kullanılır.
```

```python
df.groupby("gruplar").aggregate({"degisken1":"min", "degisken2":"max"})
```

```python
#filter() fonksiyonu, bir DataFrame'den sadece belirli satırları ya da sütunları isimlerine göre seçip yeni bir DataFrame oluşturmak için kullanılır.
def filter_func(x):
    return x["degisken1"].std()>9
```

```python
df.groupby("gruplar").filter(filter_func)
```

```python
#transform(), her satıra veya her elemana işlem uygulayıp aynı boyutta veri döndüren bir fonksiyondur.
df["degisken1"]*9
```

```python
df_a = df.iloc[:,1:3]
df_a.transform(lambda x: (x-x.mean()) / x.std())
```

```python
titanic.groupby(["sex","class"])[["survived"]].aggregate("mean")
```

```python
titanic.pivot_table("survived", index = "sex", columns = "class")
```
